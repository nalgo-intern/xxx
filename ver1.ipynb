```python
import ipywidgets as widgets
from IPython.display import display, HTML, clear_output
from transformers import pipeline

# ==========================================================
# 1. 事前学習済み感情分析モデルのロード (Hugging Face)
# ==========================================================
# 日本語に対応した感情分析パイプラインを準備
sentiment_analyzer = pipeline(
    "sentiment-analysis",
    model="jarvisx17/japanese-sentiment-analysis"
)

# ==========================================================
# 2. ジャンル別キー（抽出用キーワード）の定義
# ==========================================================
genre_keywords = {
    "グルメ・フード": {
        "味・品質": ["味", "旨", "美味", "まずい", "冷めて", "肉", "料理", "出汁", "パサパサ", "新鮮"],
        "接客・サービス": ["接客", "対応", "店員", "スタッフ", "親切", "笑顔", "愛想", "雑", "ホスピタリティ"],
        "混雑・待ち時間": ["混雑", "行列", "待ち", "並ぶ", "満席", "予約", "スムーズ", "すぐ入れた"],
        "雰囲気": ["雰囲気", "おしゃれ", "綺麗", "清潔", "静か", "うるさい", "狭い", "居心地"],
        "コスパ": ["コスパ", "高い", "安い", "リーズナブル", "価格", "値段", "割高", "お得"]
    },
    "物販店": {
        "品揃え・商品力": ["品揃え", "在庫", "商品", "限定", "ラインナップ", "可愛い", "クオリティ"],
        "接客・サービス": ["店員", "接客", "親切", "丁寧", "説明", "対応", "ラッピング"],
        "レジ・待ち時間": ["レジ", "会計", "並ぶ", "混雑", "早い", "スムーズ"],
        "価格・コスパ": ["価格", "安い", "高い", "セール", "お得", "コスパ"],
        "清潔感・店舗環境": ["清潔", "綺麗", "見やすい", "狭い", "通路", "レイアウト"]
    }
    # 他のジャンルも同様に拡張可能
}

# ==========================================================
# 3. 分析ロジック（キーワード抽出 ＋ AI感情分析）
# ==========================================================
def analyze_with_ai(text, genre):
    # 選択ジャンルのキーワード定義を取得（未定義の場合はグルメを使用）
    aspect_map = genre_keywords.get(genre, genre_keywords["グルメ・フード"])

    # テキストを文（センテンス）ごとに分割
    import re
    sentences = [s.strip() for s in re.split(r'[。！!？?\n]', text) if s.strip()]

    # 項目ごとのスコアリストを初期化
    aspect_scores = {aspect: [] for aspect in aspect_map.keys()}

    for sentence in sentences:
        # 1. センテンスがどの「項目キー」に該当するか判定
        matched_aspects = []
        for aspect, keywords in aspect_map.items():
            if any(kw in sentence for kw in keywords):
                matched_aspects.append(aspect)

        # 2. 該当する項目があれば、事前学習済みモデルで感情分析を実行
        if matched_aspects:
            # モデルの推論を実行 (例: [{'label': 'positive', 'score': 0.98}])
            result = sentiment_analyzer(sentence)[0]
            label = result['label'].lower()
            confidence = result['score'] # 確信度 (0.0 ~ 1.0)

            # Labelと確信度から 0~100 の満足度スコアに変換
            if 'pos' in label:
                # ポジティブの場合: 50点〜100点
                calculated_score = 50 + (confidence * 50)
            elif 'neg' in label:
                # ネガティブの場合: 0点〜50点
                calculated_score = 50 - (confidence * 50)
            else:
                calculated_score = 50.0

            # 該当したすべての項目にスコアを追加
            for aspect in matched_aspects:
                aspect_scores[aspect].append(calculated_score)

    # 3. 該当する文がなかった項目は標準（50点）とし、文があった項目は平均値を算出
    final_scores = {}
    for aspect, scores in aspect_scores.items():
        if scores:
            final_scores[aspect] = sum(scores) / len(scores)
        else:
            final_scores[aspect] = 50.0  # 口コミ内に言及がない場合は50%

    return final_scores

# ==========================================================
# 4. UIパーツの作成 (ipywidgets)
# ==========================================================
title_widget = widgets.HTML("<h2>📊 口コミ要素別・感情アナライザー (AIモデル版)</h2><p><b>キーワード</b>で言及項目を抽出し、<b>事前学習済みLLM/モデル</b>で感情（ポジ・ネガ）を判定します。</p>")

genre_dropdown = widgets.Dropdown(
    options=list(genre_keywords.keys()),
    value='グルメ・フード',
    description='分析ジャンル:',
    style={'description_width': 'initial'}
)

default_text = """店員さんの対応がとても温かく、特製チーズケーキが絶品でした！ただ、少し混雑していました。
休日の昼下がりはかなり混雑していて30分以上待ちました。席も狭めです。
コーヒーのこだわりが感じられて最高です。接客も丁寧でした。
決して美味しくはないですが、安いです。"""

text_area = widgets.Textarea(
    value=default_text,
    placeholder='口コミテキストを入力...',
    layout=widgets.Layout(width='100%', height='150px')
)

btn_analyze = widgets.Button(
    description='🚀 AI感情分析を実行',
    button_style='primary',
    layout=widgets.Layout(width='200px')
)

output_area = widgets.Output()

# ==========================================================
# 5. 実行イベント
# ==========================================================
def on_button_clicked(b):
    with output_area:
        clear_output()
        reviews = [r.strip() for r in text_area.value.split("\n") if r.strip()]
        selected_genre = genre_dropdown.value

        if not reviews:
            display(HTML("<span style='color:red;'>⚠️ 口コミテキストを入力してください。</span>"))
            return

        display(HTML(f"<div style='background-color:#e2e3e5; padding:8px; border-radius:4px;'>🤖 AI推論中... 対象ジャンル: <b>{selected_genre}</b></div><br>"))

        # 全口コミの平均を算出するための集計バッファ
        total_scores = {aspect: [] for aspect in genre_keywords.get(selected_genre, genre_keywords["グルメ・フード"]).keys()}

        for text in reviews:
            scores = analyze_with_ai(text, selected_genre)
            for aspect, score in scores.items():
                total_scores[aspect].append(score)

        clear_output()
        display(HTML(f"<div style='background-color:#d4edda; color:#155724; padding:8px; border-radius:4px;'>✅ 分析完了！ 対象ジャンル: <b>{selected_genre}</b> （件数: {len(reviews)}件）</div><br>"))

        display(HTML("<h3>📈 AI感情分析結果（満足度スコア）</h3>"))
        for aspect, score_list in total_scores.items():
            avg_score = sum(score_list) / len(score_list)

            p_bar = widgets.IntProgress(
                value=int(avg_score),
                min=0, max=100,
                bar_style='success' if avg_score >= 60 else ('danger' if avg_score <= 40 else 'warning'),
                layout=widgets.Layout(width='60%')
            )
            label = widgets.HTML(f"<b>{aspect}</b>: {avg_score:.1f}%")
            display(widgets.VBox([label, p_bar]))

btn_analyze.on_click(on_button_clicked)

# 配置
display(title_widget, widgets.VBox([genre_dropdown, text_area, btn_analyze]), output_area)
```

```python
# 1. 必要なライブラリのインストール
!pip install -q transformers torch ipywidgets accelerate

import json
import re
import torch
import ipywidgets as widgets
from IPython.display import display, HTML, clear_output
from transformers import AutoModelForCausalLM, AutoTokenizer, pipeline

# ==========================================================
# 1. 軽量ローカルLLMのロード (Hugging Face)
# ==========================================================
print("🤖 軽量ローカルLLMをロード中... (1分ほどかかります)")

# Google ColabのT4 GPU環境で快適に動く軽量モデル
MODEL_ID = "Qwen/Qwen2.5-3B-Instruct"

tokenizer = AutoTokenizer.from_pretrained(MODEL_ID)
model = AutoModelForCausalLM.from_pretrained(
    MODEL_ID,
    torch_dtype=torch.float16,
    device_map="auto" # GPUを自動割り当て
)

# テキスト生成パイプラインの準備
llm_pipeline = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer
)

print("✅ LLMの準備が完了しました！")

# ==========================================================
# 2. ジャンル別キー（抽出用ターゲット要素）の定義
# ==========================================================
genre_keywords = {
    "グルメ・フード": ["味・品質", "接客・サービス", "混雑・待ち時間", "雰囲気", "コスパ"],
    "物販店": ["品揃え・商品力", "接客・サービス", "レジ・待ち時間", "価格・コスパ", "清潔感・店舗環境"]
}

# ==========================================================
# 3. 分析ロジック（ローカルLLMによる一元判定）
# ==========================================================
def analyze_with_llm(text, genre):
    aspects = genre_keywords.get(genre, genre_keywords["グルメ・フード"])
    aspects_str = "、".join([f'"{a}"' for a in aspects])

    # LLMへのプロンプト作成（JSON出力を強制）
    messages = [
        {
            "role": "system",
            "content": f"""あなたは口コミの感情分析AIです。
ユーザーから入力された口コミテキストを読み、指定された項目ごとにポジティブ度（満足度）を0〜100の数値（パーセント）で評価してください。

【評価ルール】
- ポジティブな記述がある場合: 51〜100
- ネガティブな記述がある場合: 0〜49
- 言及がない（判定不能な）場合: 50
- 必ず以下の完全なJSONフォーマットでのみ回答してください。余計な解説は不要です。

出力フォーマット:
{{
  "scores": {{
    {", ".join([f'"{a}": 0' for a in aspects])}
  }}
}}"""
        },
        {
            "role": "user",
            "content": f"口コミ本文:\n「{text}」"
        }
    ]

    prompt = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)

    # 推論の実行
    outputs = llm_pipeline(
        prompt,
        max_new_tokens=256,
        do_sample=False, # 温度0に相当（再現性を高くする）
        temperature=0.0
    )

    generated_text = outputs[0]["generated_text"][len(prompt):]

    # JSONの抽出・パース処理
    try:
        # レスポンスからJSON部分を取り出す regex
        json_match = re.search(r'\{.*\}', generated_text, re.DOTALL)
        if json_match:
            data = json.loads(json_match.group(0))
            scores = data.get("scores", {})
            # スコアの型・範囲チェック
            cleaned_scores = {}
            for aspect in aspects:
                val = float(scores.get(aspect, 50))
                cleaned_scores[aspect] = max(0.0, min(100.0, val))
            return cleaned_scores
    except Exception as e:
        print(f"JSONパースエラー: {e}")

    # エラー時のフォールバック (50点均等)
    return {aspect: 50.0 for aspect in aspects}

# ==========================================================
# 4. UIパーツの作成 (ipywidgets)
# ==========================================================
title_widget = widgets.HTML("<h2>📊 口コミ要素別・感情アナライザー (軽量ローカルLLM版)</h2><p>キーワードに頼らず、<b>ローカルLLM（Qwen2.5-3B）</b>が文脈を直接解釈して要素別にスコア化（%）します。</p>")

genre_dropdown = widgets.Dropdown(
    options=list(genre_keywords.keys()),
    value='グルメ・フード',
    description='分析ジャンル:',
    style={'description_width': 'initial'}
)

default_text = """店員さんの対応がとても温かく、特製チーズケーキが絶品でした！ただ、少し混雑していました。
休日の昼下がりはかなり混雑していて30分以上待ちました。席も狭めです。
コーヒーのこだわりが感じられて最高です。接客も丁寧でした。
決して美味しくはないですが、安いです。"""

text_area = widgets.Textarea(
    value=default_text,
    placeholder='口コミテキストを入力...',
    layout=widgets.Layout(width='100%', height='150px')
)

btn_analyze = widgets.Button(
    description='🚀 LLM感情分析を実行',
    button_style='primary',
    layout=widgets.Layout(width='200px')
)

output_area = widgets.Output()

# ==========================================================
# 5. 実行イベント
# ==========================================================
def on_button_clicked(b):
    with output_area:
        clear_output()
        full_text = text_area.value.strip()
        selected_genre = genre_dropdown.value

        if not full_text:
            display(HTML("<span style='color:red;'>⚠️ 口コミテキストを入力してください。</span>"))
            return

        display(HTML(f"<div style='background-color:#e2e3e5; padding:8px; border-radius:4px;'>🤖 LLM推論中... 対象ジャンル: <b>{selected_genre}</b></div><br>"))

        # 口コミ文章全体を一括でLLMに分析させる（必要に応じて改行で分けることも可）
        scores = analyze_with_llm(full_text, selected_genre)

        clear_output()
        display(HTML(f"<div style='background-color:#d4edda; color:#155724; padding:8px; border-radius:4px;'>✅ 分析完了！ 対象ジャンル: <b>{selected_genre}</b></div><br>"))

        display(HTML("<h3>📈 AI感情分析結果（満足度スコア）</h3>"))
        for aspect, score in scores.items():
            p_bar = widgets.IntProgress(
                value=int(score),
                min=0, max=100,
                bar_style='success' if score >= 60 else ('danger' if score <= 40 else 'warning'),
                layout=widgets.Layout(width='60%')
            )
            label = widgets.HTML(f"<b>{aspect}</b>: {score:.1f}%")
            display(widgets.VBox([label, p_bar]))

btn_analyze.on_click(on_button_clicked)

# 配置
display(title_widget, widgets.VBox([genre_dropdown, text_area, btn_analyze]), output_area)
```
